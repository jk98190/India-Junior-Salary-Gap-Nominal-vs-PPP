```python
#Import the required libraries 
import pandas as pd 
import numpy as np 
import sys
```

```python
# Load the dataset 
skill_salary= pd.read_csv('/Users/jahaankachwala/Downloads/Projects /Global Salary Data/skill_salary_percentiles_global_sample.csv')
india_salary = pd.read_csv('/Users/jahaankachwala/Downloads/Projects /Global Salary Data/india_salary_benchmarks_2026_sample.csv')
economic_indicators = pd.read_csv(
    '/Users/jahaankachwala/Downloads/API_NY.GDP.MKTP.CD_DS2_en_csv_v2_512/API_NY.GDP.MKTP.CD_DS2_en_csv_v2_512.csv',
    skiprows=4,  # World Bank files have metadata rows before the real header
    encoding='utf-8-sig',
    engine='python'
)
```

```python
#File cleaning and preprocessing for skill_salary dataset 
                     
skill_salary.columns = skill_salary.columns.str.strip().str.lower().str.replace(" ", "_")  # tidy headers
skill_salary = skill_salary.apply(lambda c: c.str.strip() if c.dtype == "object" else c)   # trim text cells
skill_salary = skill_salary.drop_duplicates().dropna(how="all")                            # dupes + blank rows

skill_salary.to_csv("skill_salary_clean.csv", index=False)              # original untouched
```

```python
#File cleaning and preprocessing for India_salary dataset 

india_salary.columns = india_salary.columns.str.strip().str.lower().str.replace(" ", "_")  # tidy headers
india_salary = india_salary.apply(lambda c: c.str.strip() if c.dtype == "object" else c)   # trim text cells
india_salary = india_salary.drop_duplicates().dropna(how="all")                            # dupes + blank rows

india_salary.to_csv("india_salary_clean.csv", index=False)              # original untouched
```

```python
#File cleaning and preprocessing for economic_indicators Dataset 
economic_indicators.columns = economic_indicators.columns.str.strip().str.lower().str.replace(" ", "_")  # tidy headers
economic_indicators = economic_indicators.apply(lambda c: c.str.strip() if c.dtype == "object" else c)   # trim text cells
economic_indicators = economic_indicators.drop_duplicates().dropna(how="all")                            # dupes + blank rows

economic_indicators.to_csv("economic_indicators_clean.csv", index=False)              # original untouched
```
